## 🔄 Mutable vs Immutable

| Type | Mutable? | Example |
|------|----------|---------|
| **List** | ✅ Yes | `[1, 2, 3]` |
| **Dict** | ✅ Yes | `{"a": 1}` |
| **Set** | ✅ Yes | `{1, 2, 3}` |
| **Int** | ❌ No | `5` |
| **Float** | ❌ No | `3.14` |
| **Str** | ❌ No | `"hello"` |
| **Tuple** | ❌ No | `(1, 2, 3)` |

## 🟢 Mutable — Can Modify After Creation

Two variables pointing to same list → both see changes!

````python
list_a: list[int] = [1,2,3,4,5,6,999]
list_b = list_a
list_a.append(77)
print("list_b:", list_b)
````

# Immutable 
* can't modified after creation

## 🔴 Immutable — Cannot Modify After Creation

Each change creates a **new** object in memory.

````python
num_a:int= 5
num_b = num_a
num_a  += 1
print("num_b:", num_b)
print("num_a:", num_a)
````

## 🔍 Memory Addressing with `id()`

`id(obj)` shows the **memory address** of an object.

- Same `id` = same object
- Different `id` = different object

# Adddresing changes without function caling

````python
print("hello word")
````

````python
x = 10
print("before modification", x, id(x))
x += 1
print("After modification", x, id(x))
````

## 🔗 Pass by Reference vs Pass by Value

When you pass a variable to a function:

- **Immutable** (int, str, tuple) → passed by **value** (copy)
- **Mutable** (list, dict, set) → passed by **reference** (shared)

````python
a: int = 5
def abc(num1:int)-> None:
    num1 = 6
    print(f"Num1 value is {num1}")
abc(a)

print(a)    
````

````python
a: list[int] = [1,2,3,4,5]
def abc(num1:int)-> None:
    num1.append(200)
    print(f"Num1 value is {num1}")
abc(a)
print(a)    
````

````python
a: list[int] = [1,2,3,4,5]
def abc(num1:int)-> None:
    num1 = [7]
    num1.append(200)
    print(f"Num1 value is {num1}")
abc(a)
print(a)    
````

````python
a: int = 5
def abc(num1:int)-> None:
    num1 = 6
    print(f"Num1 value is {num1} address {id(a)}")
abc(a)

print(f"{a}  address", id(a))    
````

````python
x: int = 7
b = x
print(f"X value is {x} and address is {id(x)}")

print(f"b value is {b} and address is {id(x)}")

b =200

print(f"b value is {b} and address is {id(x)}")
````

## 🛡️ Handle Errors with `try` / `except`

**Syntax:**
```python
try:
    # risky code
except (ErrorClass1, ErrorClass2):
    # runs if error occurs
else:
    # runs if NO error
finally:
    # ALWAYS runs

````python
print("logic1")
print("logic2")

li:  list[int] = [1,2,3,4,5]
try:
    print(7/0)
    print(li[0])
except (ZeroDivisionError, IndexError,NameError):
    print("There is ZeroDivisionError  a error")  
print("logic1")
print("logic2")      
````

## 📋 Complete `try-except` Structure

```python
try:
    # code that might fail
except ValueError:
    # runs if ValueError
except (TypeError, KeyError):
    # runs for other errors
else:
    # runs if NO error
finally:
    # ALWAYS runs (cleanup)

## 🎓 Key Takeaways

- ✅ **Mutable** (list, dict, set) — can modify after creation
- ✅ **Immutable** (int, str, tuple) — cannot modify
- ✅ `id()` shows memory address
- ✅ Mutable objects share memory (references)
- ✅ Immutable objects create new memory on change
- ✅ Passing to functions: mutable = shared, immutable = copy
- ✅ Common errors: `ValueError`, `TypeError`, `KeyError`, `IndexError`, `ZeroDivisionError`
- ✅ `try`/`except` prevents crashes
- ✅ One `except` can handle multiple errors
- ✅ Code after `try-except` continues normally
