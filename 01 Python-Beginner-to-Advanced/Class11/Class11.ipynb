# Handle run time error

'''
* try: 
  *   logic
* except (Error_class. error-class2)
   *  if error accured then run this block 
* else: 
  *   if error not accured
* finally:
      *   always run            
'''

```python
print(7/0)
```

```python

try:
    print(7/0)
except (ZeroDivisionError):
    print("There is   a error")    
```

```python
print("logic1")
print("logic2")
try:
    print(7/0)
except (ZeroDivisionError):
    print("There is   a error")    
```

```python
print("logic1")
print("logic2")
try:
    print(7/0)
except (ZeroDivisionError):
    print("There is ZeroDivisionError  a error")  
print("logic1")
print("logic2")      
```
