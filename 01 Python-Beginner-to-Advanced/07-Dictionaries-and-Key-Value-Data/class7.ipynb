## 📖 What is a Dictionary?

A **dictionary** stores data as **key : value** pairs.

| Concept | Meaning |
|---------|---------|
| **key** | Unique identifier (like index in list) |
| **value** | Data associated with key |
| **items** | All key-value pairs |

**Syntax:**
```python
data = {"key1": "value1", "key2": "value2"}

````python
data : list[str] = ['muhammad', 'waleed', 'khawar']
print(data)
print(data[1])
````

````python
from typing import Dict
import pprint

data: Dict[str, str] = {
    "fname": "Muhhmad",
    "name": "eale",
    "education": "BSCS",
}

pprint.pprint(data)
````

## 🧩 Set vs List vs Dict

| Type | Syntax | Duplicates | Order |
|------|--------|-----------|-------|
| **List** | `[1, 2, 3]` | ✅ Yes | ✅ Ordered |
| **Set** | `{1, 2, 3}` | ❌ No | ❌ Unordered |
| **Dict** | `{"a": 1}` | ❌ Keys unique | ✅ Ordered* |

*Python 3.7+ keeps insertion order

````python
from  typing import Dict, Union, Optional
import pprint

Key = Union[int, str]
Value = Union[int, str, list, dict, tuple, set]

data: Dict[Key, Value] = {"fname": "Muhhmad", 
                        "1": "eale", 
                        "education": "BSCS"}
pprint.pprint(data)
````

## 🎁 Nested Dictionaries

Dictionaries can contain lists, dicts, tuples — anything!

````python
from typing import Dict, Union
import pprint

Key = Union[int, str]
Value = Union[int, str, list, tuple, set]

data: Dict[Key, Value] = {
    "fname": "Muhhmad",
    "name": "eale",
    "education": "BSCS",
    "abc": [1, 2, 5],
    "cde": {"a": 1, "b": 2},
}

print(data)
print(data["fname"])
print(data["abc"])
print(data["cde"])
print(data["cde"]["b"])
````

## 📋 All Dictionary Methods

| Method | Purpose |
|--------|---------|
| `keys()` | All keys |
| `values()` | All values |
| `items()` | All key-value pairs |
| `get(k, default)` | Safe access |
| `pop(k)` | Remove & return value |
| `popitem()` | Remove last item |
| `clear()` | Remove all |
| `copy()` | Shallow copy |
| `update(d)` | Merge another dict |
| `setdefault(k, v)` | Get or set default |
| `fromkeys(keys, val)` | Create dict from keys |

## 🎓 Key Takeaways

- ✅ Dict = `{key: value}` pairs

- ✅ `.get()` safer than `[]`
