# 📖 What is Encapsulation?

**Encapsulation** is one of the main concepts of Object-Oriented Programming.

> **Keeping data and the methods that work with that data together inside a class, and controlling how that data can be accessed or changed.**

In simple words:

> 🔐 **Encapsulation = Protect data + Control access + Keep related behavior together**

### Bank account example

```text
Owner
Balance
Deposit
Withdraw
```

The balance is important. We should **not** allow this:

```python
account.balance = -500000
```

Instead we control changes through methods:

```python
account.deposit(5000)
account.withdraw(2000)
```

The class checks if the operation is valid **before** changing the data.

# 🎯 Why Do We Need Encapsulation?

First, let's see the **problem** — a class with no protection:

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance


account = BankAccount(50000)

account.balance = -100000   # Nothing stops us!

print(account.balance)
````

😬 The balance became **-100000**. That may not be valid for a real application.

Now the **solution** — with Encapsulation, the class controls the balance:

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(50000)

account.deposit(5000)
account.deposit(-999)      # rejected silently

print(account.get_balance())
````

# 🧠 Main Goals of Encapsulation

- 🔐 Protect important data
- 🛡️ Prevent invalid changes
- ✅ Validate input
- 🎯 Control access to internal data
- 🧹 Keep code organized
- 🔄 Make code easier to change later
- 🧩 Keep related data and behavior together
- 🏗️ Build maintainable applications
- 🐛 Reduce accidental modification of object state

# 📊 Access Levels in Python

| Type | Syntax | Meaning |
|---|---|---|
| Public | `name` | Normal public member |
| Protected | `_name` | Intended for internal/subclass use |
| Private | `__name` | Name-mangled member |

````python
class Person:
    def __init__(self):
        self.name = "Waleed"       # Public
        self._age = 23             # Protected
        self.__salary = 50000      # Private-style


p = Person()
print(p.name)
print(p._age)
````

# 1️⃣ Public Members

A **public member** can be accessed normally from outside the class. It does **not** start with an underscore.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Waleed", 23)

print(student.name)
print(student.age)
````

Public members can also be changed freely:

````python
student.name = "Ali"
student.age = 24

print(student.name, student.age)
````

### 📌 Public member syntax

```python
self.name
self.age
self.salary
self.email
```

Use public members when the value is **safe to expose directly**.

# 2️⃣ Protected Members

A protected-style member starts with **one underscore**: `_name`

The underscore says: *"This is intended for internal use or use by child classes."*

````python
class Person:
    def __init__(self, name):
        self._name = name


class Student(Person):
    def show_name(self):
        print(self._name)     # child class can use it


student = Student("Waleed")

student.show_name()
````

### ⚠️ Important point

Python does **not** strictly block outside access to `_name`. Look:

````python
print(student._name)   # still works!
````

The single underscore is only a **developer convention**. It tells other developers: *"Please treat this as an internal detail."*

# 3️⃣ Private Members

A private-style member starts with **two underscores**: `__name`

Python applies **name mangling** to it.

````python
class Person:
    def __init__(self, name):
        self.__name = name

    def show_name(self):
        print(self.__name)


person = Person("Waleed")

person.show_name()
````

Trying to access it directly from outside fails:

````python
try:
    print(person.__name)
except AttributeError as e:
    print("Error:", e)
````

### 🔐 Why use `__name`?

- Makes direct access less convenient
- Helps avoid accidental name conflicts (especially with inheritance)
- Good for internal implementation details

> ⚠️ Python does **not** provide absolute private access like some other languages.

# 4️⃣ Name Mangling

When Python sees `self.__name` inside a class called `Person`, it internally renames it to approximately:

```python
self._Person__name
```

This is called **Name Mangling**.

````python
class Person:
    def __init__(self):
        self.__name = "Waleed"


person = Person()

print(person._Person__name)   # works - so it's NOT truly private
print(person.__dict__)        # see the real stored name
````

### 🧠 Why does name mangling exist?

- Avoid accidental name conflicts
- Protect internal implementation details
- Make inheritance safer when parent and child use similar names

It is **not designed as a security mechanism**.

# 📊 Public vs Protected vs Private

| Feature | Public | Protected | Private |
|---|---|---|---|
| Syntax | `name` | `_name` | `__name` |
| Direct access | ✅ Normal | ⚠️ Possible | ⚠️ Name-mangled |
| Main purpose | Public data | Internal/subclass convention | Reduce accidental access/conflicts |
| Strictly enforced? | No | No | No |
| Name mangling | ❌ | ❌ | ✅ |

# 5️⃣ Getter Method

A **getter** is a method used to **read** a value.

````python
class Student:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name


student = Student("Waleed")

print(student.get_name())
````

### 🎯 Why use getters?

- Read internal data
- Hide the internal attribute name
- Change internal implementation later
- Add extra logic if necessary

The user does not need to know **how** the value is stored.

# 6️⃣ Setter Method

A **setter** is a method used to **change** a value (usually with validation).

````python
class Student:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")


student = Student(23)

print(student.get_age())

student.set_age(24)

print(student.get_age())
````

### ❌ Setter validation
The setter can reject invalid values:

````python
student.set_age(-5)

print(student.get_age())   # still 24
````

# 7️⃣ Getter + Setter Together

They are commonly used together.

````python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Salary cannot be negative")


employee = Employee(50000)

print(employee.get_salary())

employee.set_salary(60000)

print(employee.get_salary())

employee.set_salary(-5000)
````

### 📊 Getter vs Setter

| Method | Purpose |
|---|---|
| Getter | Read a value |
| Setter | Change a value |
| Getter + Setter | Controlled read/write access |

# 8️⃣ The `@property` Decorator

`@property` makes a method behave like an **attribute**.

Without property: `student.get_name()`

With property: `student.name`  ← cleaner!

````python
class Student:
    def __init__(self, name):
        self.__name = name

    @property
    def name(self):
        return self.__name


student = Student("Waleed")

print(student.name)     # no brackets!
````

### 🧠 What is actually happening?

When we write `student.name`, Python **runs the property method** behind the scenes:

```python
@property
def name(self):
    return self.__name
```

So the method acts like an attribute when reading.

# 9️⃣ `@property.setter`

Combine `@property` with a setter to control updates.

````python
class Student:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value > 0:
            self.__age = value
        else:
            print("Age must be positive")


student = Student(23)

print(student.age)

student.age = 24      # looks normal, but runs the setter

print(student.age)

student.age = -10     # rejected

print(student.age)
````

### 🔄 Getter/Setter methods vs `@property`

| Approach | Reading | Updating |
|---|---|---|
| Getter/Setter | `get_age()` | `set_age(24)` |
| Property | `age` | `age = 24` |

Both give controlled access. Property looks cleaner.

# 🔟 Data Validation

**Validation** = checking whether data is acceptable *before* storing it.

Notice the trick: in `__init__` we use `self.price = price` (not `self.__price`) so the setter validation also runs when the object is created.

````python
class Product:
    def __init__(self, price):
        self.__price = 0
        self.price = price       # uses the setter

    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, value):
        if value >= 0:
            self.__price = value
        else:
            print("Price cannot be negative")


product = Product(1500)

print(product.price)

product.price = -500

print(product.price)
````

# 🧪 Types of Validation

Encapsulation can validate many kinds of data. Here are quick examples:

````python
value = 10
age = 25
salary = 50000
price = 100
name = "Waleed"
email = "waleed@gmail.com"

print(value >= 0)            # Numbers
print(0 < age <= 120)        # Age
print(salary >= 0)           # Salary
print(price > 0)             # Price
print(len(name) >= 3)        # String length
print("@" in email)          # Email (simple check)
````

The exact validation depends on your application.

# 1️⃣1️⃣ Read-Only Property

A property is **read-only** if it has a getter but **no setter**.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance


account = BankAccount(50000)

print(account.balance)

try:
    account.balance = 60000
except AttributeError as e:
    print("Error:", e)
````

### 🔒 Why use read-only properties?

Useful when a value should be exposed but **not changed** by outside code:

- Employee ID
- Account number
- Created-at timestamp
- Calculated total
- System-generated identifiers

# 1️⃣2️⃣ Encapsulation with Methods

Encapsulation is **not only about private variables**. A class can protect its state by allowing changes only through controlled methods.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance


account = BankAccount("Waleed", 50000)

print(account.get_balance())

account.deposit(10000)

account.withdraw(15000)

print(account.get_balance())
````

# 🏦 1️⃣3️⃣ Real-World Example — Bank Account

The internal `self.__balance` should not be randomly modified. We provide `deposit()`, `withdraw()` and a read-only `balance` property. **The class controls the rules.**

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit must be greater than 0")
            return

        self.__balance += amount
        print(f"Deposited: {amount}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal must be greater than 0")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")


account = BankAccount("Waleed", 50000)

print("Owner:", account.owner)
print("Balance:", account.balance)

account.deposit(10000)
account.withdraw(15000)
account.withdraw(100000)

print("Final Balance:", account.balance)
````

# 1️⃣4️⃣ Encapsulation with Inheritance

A child class can use **protected** data from its parent.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary


class Manager(Employee):
    def show_salary(self):
        print(f"{self.name}'s salary: {self._salary}")


manager = Manager("Waleed", 100000)

manager.show_salary()
````

Here `_salary` is protected-style data, and the child class can access it.

# 1️⃣5️⃣ Private Members and Inheritance

Private-style attributes are name-mangled with the **parent's class name**, so a child class cannot use them directly. Let's see the problem first:

````python
class Person:
    def __init__(self, name):
        self.__name = name


class Student(Person):
    def show_name(self):
        print(self.__name)     # tries self._Student__name


student = Student("Waleed")

try:
    student.show_name()
except AttributeError as e:
    print("Error:", e)
````

✅ The solution: the child uses the parent's **public method** instead.

````python
class Person:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name


class Student(Person):
    def show_name(self):
        print(self.get_name())


student = Student("Waleed")

student.show_name()
````

This keeps the implementation **controlled by the parent class**.

# 1️⃣6️⃣ Encapsulation with `super()`

`super()` lets a child class call the parent class. It is very useful when the parent handles initialization.

````python
class Person:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name


class Student(Person):
    def __init__(self, name, roll_number):
        super().__init__(name)
        self.roll_number = roll_number

    def show_student(self):
        print("Name:", self.get_name())
        print("Roll Number:", self.roll_number)


student = Student("Waleed", 101)

student.show_student()
````

The parent controls its private `__name`, while the child adds its own data.

# 👨‍💼 1️⃣7️⃣ Real-World Example — Employee Salary

We can prevent invalid salary values.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = 0
        self.salary = salary

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, value):
        if value >= 0:
            self.__salary = value
        else:
            print("Salary cannot be negative")


employee = Employee("Waleed", 75000)

print(employee.salary)

employee.salary = 80000

print(employee.salary)

employee.salary = -5000

print(employee.salary)
````

# 🛒 1️⃣8️⃣ Real-World Example — Product Price

E-commerce applications often need price validation.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.__price = 0
        self.price = price

    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, value):
        if value >= 0:
            self.__price = value
        else:
            print("Price cannot be negative")


product = Product("Laptop", 120000)

print(product.name)
print(product.price)

product.price = 130000

print(product.price)
````

# 🧮 1️⃣9️⃣ Encapsulation with Calculated Properties

Properties can also **calculate** values. `area` is not stored — it is calculated every time you access it.

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height


rectangle = Rectangle(10, 5)

print(rectangle.area)

rectangle.width = 20      # change width...

print(rectangle.area)     # ...area updates automatically
````

# 🔐 Encapsulation is More Than Private Variables

A common beginner mistake:

> ❌ "Encapsulation means making variables private."

That is only **part** of the idea. Encapsulation is mainly about:

```text
Data
  +
Related Methods
  +
Controlled Access
  +
Validation
```

Example structure of a bank account:

```text
BankAccount
   private balance
        ↓
   deposit()
   withdraw()
   balance property
```

The important part: **the class controls how its internal state is used.**

# 🔒 Encapsulation vs Data Hiding

They are related but **not exactly the same**.

**Encapsulation:** Combining data and related behavior inside a class and controlling access to that state.

**Data Hiding:** Preventing or discouraging direct access to internal implementation details.

```text
Encapsulation  →  Organize + Control
Data Hiding    →  Limit/Discourage Direct Access
```

Data hiding can be **one part** of Encapsulation.

| Concept | Main Focus |
|---|---|
| Encapsulation | Combine data + behavior and control access |
| Data Hiding | Hide or restrict internal implementation details |

# 🐍 Important Python Point

Python does **not** have strict access modifiers (`public`, `private`, `protected`) like some other languages. Instead it uses conventions and language features:

- `name` → normal public access
- `_name` → convention meaning "treat this as internal"
- `__name` → triggers name mangling
- `@property` → controlled attribute-like access

# 🧠 Common Encapsulation Patterns

Let's see the 5 patterns in real, runnable code.

## Pattern 1 — Private Data + Getter

````python
class User:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name


print(User("Waleed").get_name())
````

## Pattern 2 — Private Data + Getter + Setter

````python
class User:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age


u = User(23)
u.set_age(30)
print(u.get_age())
````

## Pattern 3 — Property

````python
class User:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age


print(User(23).age)
````

## Pattern 4 — Property + Setter

````python
class User:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value > 0:
            self.__age = value


u = User(23)
u.age = 40
print(u.age)
````

## Pattern 5 — Controlled Methods

````python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance


b = BankAccount(1000)
b.deposit(500)
b.withdraw(200)
print(b.get_balance())
````

# ⚠️ Common Mistakes

## Mistake 1 — Thinking `_name` is truly private
`self._name` is **not** truly private. It is a convention.

## Mistake 2 — Thinking `__name` is completely inaccessible
`self.__name` uses name mangling. It is **not** a security mechanism.

## Mistake 3 — Making Everything Private
Not every variable needs to be private. Use access control only when it gives a real benefit.

## Mistake 4 — Using Getters and Setters Without a Reason
Sometimes a plain public attribute is perfectly fine:

````python
class Student:
    def __init__(self, name):
        self.name = name      # simple and fine - no special rules needed


s = Student("Ali")
s.name = "Waleed"
print(s.name)
````

## Mistake 5 — Forgetting Validation
If a setter exists, think about whether invalid values should be rejected, for example `if salary >= 0:`

A setter **without** validation adds no protection at all.

# 🎯 Encapsulation in a Real Application

Imagine an e-commerce backend with an `Order` class:

```text
Order
│
├── order_id
├── customer
├── items
├── total
├── status
│
├── add_item()
├── remove_item()
├── calculate_total()
└── update_status()
```

Instead of `order.total = -500`, changes go through class methods. Here is a small working version:

````python
class Order:
    def __init__(self, order_id, customer):
        self.__order_id = order_id
        self.customer = customer
        self.__items = []
        self.__status = "pending"

    @property
    def order_id(self):
        return self.__order_id

    @property
    def status(self):
        return self.__status

    @property
    def total(self):
        return sum(price for name, price in self.__items)

    def add_item(self, name, price):
        if price < 0:
            print("Price cannot be negative")
        else:
            self.__items.append((name, price))

    def mark_as_paid(self):
        if self.total > 0:
            self.__status = "paid"
        else:
            print("Cannot pay for an empty order")


order = Order(1, "Waleed")

order.add_item("Laptop", 120000)
order.add_item("Mouse", 2000)
order.add_item("Bad item", -50)

print("Total:", order.total)
print("Status:", order.status)

order.mark_as_paid()

print("Status:", order.status)

try:
    order.total = -500
except AttributeError as e:
    print("Cannot change total directly:", e)
````

# 🎤 Interview Questions

Great for revision. Try answering out loud before reading the answer!

### Q1. What is Encapsulation?

**Answer:**

Encapsulation is the OOP concept of keeping data and related methods together inside a class while controlling how the data is accessed or modified.

### Q2. What is the difference between `_name` and `__name`?

**Answer:**

`_name` is a protected-style convention, while `__name` triggers name mangling.

### Q3. Is `__name` truly private in Python?

**Answer:**

No. Python uses name mangling rather than strict private access.

### Q4. What is a getter?

**Answer:**

A getter is a method or property used to read an object's internal data.

### Q5. What is a setter?

**Answer:**

A setter is a method or property used to change an object's data, often with validation.

### Q6. What is `@property`?

**Answer:**

`@property` allows a method to be accessed like an attribute.

### Q7. Why is Encapsulation useful?

**Answer:**

It helps control data access, validate values, prevent accidental changes, organize code, and make applications easier to maintain.

### Q8. Is Encapsulation only about private variables?

**Answer:**

No. Encapsulation is broader. It includes combining data and behavior and controlling how an object's internal state is used.
