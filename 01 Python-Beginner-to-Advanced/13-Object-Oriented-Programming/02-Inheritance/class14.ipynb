## 📖 What is Inheritance?

Inheritance allows one class to use the attributes and methods of another class.

The class that provides the features is called the **parent class**.

The class that receives the features is called the **child class**.

Inheritance helps us reuse code instead of writing the same code again.

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    pass


dog1 = Dog()

dog1.eat()
````

## 👨‍👦 Parent Class and Child Class

- `Animal` is the parent class.
- `Dog` is the child class.
- `Dog` inherits the `eat()` method from `Animal`.

The child class can use features from the parent class.

# 1️⃣ Single Inheritance

**Single inheritance** means one child class inherits from one parent class.

Example:

```text
Animal
   ↓
  Dog

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog1 = Dog()

dog1.eat()
dog1.bark()
````

### 🧠 Explanation

`Dog` inherits from `Animal`.

So the `Dog` object can use:

- `eat()` from `Animal`.
- `bark()` from `Dog`.

# 2️⃣ Multilevel Inheritance

**Multilevel inheritance** means inheritance happens across multiple levels.

Example:

```text
Animal
   ↓
  Dog
   ↓
 Puppy

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Puppy(Dog):
    def play(self):
        print("Puppy is playing")


puppy1 = Puppy()

puppy1.eat()
puppy1.bark()
puppy1.play()
````

# 3️⃣ Multiple Inheritance

**Multiple inheritance** means one child class inherits from more than one parent class.

Example:

```text
Father      Mother
    \        /
      Child

````python
class Father:
    def work(self):
        print("Father is working")


class Mother:
    def care(self):
        print("Mother is caring")


class Child(Father, Mother):
    pass


child1 = Child()

child1.work()
child1.care()
````

# 4️⃣ Hierarchical Inheritance

**Hierarchical inheritance** means multiple child classes inherit from the same parent class.

Example:

```text
       Animal
       /    \
     Dog    Cat

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Cat(Animal):
    def meow(self):
        print("Cat is meowing")


dog1 = Dog()
cat1 = Cat()

dog1.eat()
dog1.bark()

cat1.eat()
cat1.meow()
````

# 5️⃣ Hybrid Inheritance

**Hybrid inheritance** is a combination of two or more inheritance types.

Example:

```text
        Animal
       /      \
     Dog      Cat
       \      /
        Pet

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Cat(Animal):
    def meow(self):
        print("Cat is meowing")


class Pet(Dog, Cat):
    pass


pet1 = Pet()

pet1.eat()
pet1.bark()
pet1.meow()
````

# 6️⃣ `super()` Function

The `super()` function is used to access functionality from a parent class.

It is commonly used when a child class needs to use the parent's constructor or method.

````python
class Animal:
    def __init__(self, name):
        self.name = name


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed


dog1 = Dog("Buddy", "German Shepherd")

print(dog1.name)
print(dog1.breed)
````

### 🧠 Why use `super()`?

`super()` allows the child class to use the parent class functionality.

This avoids repeating the parent code.

# 7️⃣ Method Overriding

**Method overriding** happens when a child class creates a method with the same name as a method in the parent class.

The child version replaces the parent version when called through the child object.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


dog1 = Dog()

dog1.sound()
````

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks")


dog1 = Dog()

dog1.sound()
````

# 9️⃣ Constructor Inheritance

If a child class does not have its own `__init__()` method, it can use the parent's `__init__()` method.

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    pass


student1 = Student("Waleed")

print(student1.name)
````

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    pass


student1 = Student("Waleed")

print(student1.name)
````

# 1️⃣1️⃣ Inheriting Attributes and Methods

A child class can use:

- Parent attributes.
- Parent methods.
- Its own attributes.
- Its own methods.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}")


class Student(Person):
    def study(self):
        print(f"{self.name} is studying")


student1 = Student("Waleed", 23)

print(student1.age)
student1.introduce()
student1.study()
````

# 1️⃣4️⃣ `isinstance()`

`isinstance()` checks whether an object belongs to a class.

It also returns `True` for parent classes of that object's class.

````python
class Animal:
    pass


class Dog(Animal):
    pass


dog1 = Dog()

print(isinstance(dog1, Dog))
print(isinstance(dog1, Animal))
````

# 1️⃣5️⃣ `issubclass()`

`issubclass()` checks whether one class is a child of another class.

````python
class Animal:
    pass


class Dog(Animal):
    pass


print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))
````

# 1️⃣7️⃣ Private Members

Python uses two underscores `__` for private-style attributes.

Python applies **name mangling** to these names.

A child class normally cannot access the private attribute directly.

````python
class Person:
    def __init__(self):
        self.__name = "Waleed"

    def show_name(self):
        print(self.__name)


person1 = Person()

person1.show_name()
````

# 1️⃣8️⃣ Abstract Classes and Inheritance

An **abstract class** provides a common structure for child classes.

Python provides abstract classes using the `abc` module.

An abstract method is a method that child classes are expected to implement.

````python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog1 = Dog()

dog1.sound()
````

# 🎬 Real-World Example — Employee System

Imagine a company has different employees.

All employees can have:

- Name
- Salary
- Basic information

But different employees can have different actions.

For example:

- Developer → writes code
- Manager → manages a team

We can put the common features in an `Employee` parent class and create child classes for specific roles.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")


class Developer(Employee):
    def code(self):
        print(f"{self.name} is writing code")


class Manager(Employee):
    def manage(self):
        print(f"{self.name} is managing the team")


developer1 = Developer("Waleed", 100000)
manager1 = Manager("Ali", 150000)

developer1.show_info()
developer1.code()

print()

manager1.show_info()
manager1.manage()
````

# 📊 Types of Inheritance

| Type | Meaning |
|---|---|
| Single | One parent → one child |
| Multilevel | Parent → child → grandchild |
| Multiple | Multiple parents → one child |
| Hierarchical | One parent → multiple children |
| Hybrid | Combination of inheritance types |
