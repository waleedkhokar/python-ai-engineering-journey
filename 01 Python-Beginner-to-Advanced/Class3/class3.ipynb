## 🔢 Python Operator Groups

Python divides operators into **7 groups**:

| # | Group | Purpose |
|---|-------|---------|
| 1 | Arithmetic | Math operations |
| 2 | Assignment | Assign values |
| 3 | Comparison | Compare values |
| 4 | Logical | Combine conditions |
| 5 | Identity | Check same object |
| 6 | Membership | Check in collection |
| 7 | Bitwise | Bit-level operations |

## 📝 2. Assignment Operators

```python
a : int = 7
print(a)     # 7

a += 5       # a = a + 5 = 12
print(a)     # 12

a -= 5       # a = a - 5 = 7
print(a)     # 7

a *= 5       # a = a * 5 = 35
print(a)     # 35

a /= 5       # a = a / 5 = 7.0
print(a)     # 7.0

a %= 5       # a = a % 5
print(a)     # 2.0

a //= 5
print(a)     # 0.0
