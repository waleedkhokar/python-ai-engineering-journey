# 🧠 NumPy Array

The most important concept in NumPy is the **array**.

A NumPy array is a collection of values stored in a structured numerical format.

````python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
print(type(numbers))
````

---
# 📊 Python List vs NumPy Array

Python list:

```python
numbers = [10, 20, 30, 40, 50]
```

NumPy array:

```python
numbers = np.array([10, 20, 30, 40, 50])
```

NumPy arrays are designed for numerical operations and can perform many calculations more efficiently and conveniently than normal Python lists.

````python
py_list = [10, 20, 30, 40, 50]
np_array = np.array([10, 20, 30, 40, 50])

# Multiplying a list REPEATS it; multiplying an array does MATH on each element
print("List  * 2 :", py_list * 2)
print("Array * 2 :", np_array * 2)
````

---
# 📐 Dimensions of Arrays

NumPy arrays can have different dimensions.

## 1D Array

````python
a1 = np.array([1, 2, 3, 4])
print(a1)
````

## 2D Array

This can be thought of as:

```text
Rows × Columns

2 × 3
```

````python
a2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(a2)
````

## 3D Array

3D arrays are useful when working with more complex numerical structures such as image data.

````python
a3 = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])
print(a3)
print("Shape:", a3.shape)
````

# 🔍 Important Array Properties

NumPy provides useful properties for understanding an array.

````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("ndim    :", data.ndim)
print("shape   :", data.shape)
print("size    :", data.size)
print("dtype   :", data.dtype)
print("itemsize:", data.itemsize)
````

| Property   | Meaning                     |
| ---------- | --------------------------- |
| `ndim`     | Number of dimensions        |
| `shape`    | Size of each dimension      |
| `size`     | Total number of elements    |
| `dtype`    | Data type of elements       |
| `itemsize` | Memory size of each element |

# 🎯 Indexing

NumPy supports indexing similar to Python lists.

````python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[0])
print(numbers[2])
print(numbers[-1])
````

Output:

### Multi-dimensional indexing

````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data[1, 2])   # row 1, column 2 -> 60
print(data[0])      # first row
print(data[:, 1])   # second column
````

# ✂️ Slicing

NumPy supports slicing.

````python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[1:4])
print(numbers[::2])    # step slicing
print(numbers[::-1])   # reverse slicing
````

Output of `numbers[1:4]`:

```text
[20 30 40]
```

For a 2D array:

````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data[0:2, 1:3])
````

# ➕ Mathematical Operations

NumPy makes numerical calculations simple.

````python
numbers = np.array([10, 20, 30, 40])

print(numbers + 10)
print(numbers - 5)
print(numbers * 2)
print(numbers / 10)
````

Output:

```text
[20 30 40 50]
[ 5 15 25 35]
[20 40 60 80]
[1. 2. 3. 4.]
```

### More math functions

````python
x = np.array([1, 4, 9, 16])

print(x ** 2)          # power
print(x % 3)           # modulus
print(np.abs(np.array([-3, -1, 2])))
print(np.sqrt(x))
print(np.exp(np.array([0, 1, 2])))
print(np.log(np.array([1, np.e, np.e**2])))
print(np.sin(np.array([0, np.pi/2, np.pi])))
````

# 📊 Statistical Functions

NumPy provides many statistical functions. These operations are extremely important in Data Science.

````python
data = np.array([10, 20, 30, 40, 50])

print("Sum      :", np.sum(data))
print("Mean     :", np.mean(data))
print("Median   :", np.median(data))
print("Min      :", np.min(data))
print("Max      :", np.max(data))
print("Std Dev  :", np.std(data))
print("Variance :", np.var(data))
print("75th pct :", np.percentile(data, 75))
print("Argmin   :", np.argmin(data))
print("Argmax   :", np.argmax(data))
````

# 🔁 Operations Between Arrays

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)
print(b - a)
print(a * b)
print(b / a)

# Comparison and Boolean operations
print(a > 1)
print(np.logical_and(a > 1, b < 30))
````

# 🎯 Boolean Filtering

NumPy can filter data using conditions. This concept becomes very important later when working with Pandas and datasets.

````python
numbers = np.array([10, 20, 30, 40, 50])

result = numbers[numbers > 25]

print(result)
````

Output:

```text
[30 40 50]
```

### Multiple conditions

````python
mask = (numbers > 15) & (numbers < 45)
print(mask)
print(numbers[mask])
````

---
# 🔎 Searching and Sorting

````python
arr = np.array([40, 10, 30, 20, 50])

print(np.sort(arr))
print(np.where(arr > 25))              # indexes where condition is True
print(np.where(arr > 25, "big", "small"))

m = np.array([[3, 1, 2], [9, 7, 8]])
print(np.sort(m, axis=1))              # sort each row
````

# 📐 Reshaping

An array can be reshaped into another structure. The number of elements must remain the same.

````python
numbers = np.arange(1, 7)

matrix = numbers.reshape(2, 3)

print(matrix)
````

Output:

```text
[[1 2 3]
 [4 5 6]]
```

### Flattening

````python
print(matrix.flatten())   # always returns a copy
print(matrix.ravel())     # returns a view when possible
````

---
# 🔄 Transpose

Transpose changes rows into columns and columns into rows.

````python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix.T)
````

Output:

```text
[[1 4]
 [2 5]
 [3 6]]
```

---
# 📡 Broadcasting

Broadcasting allows NumPy to perform operations between arrays with compatible shapes.

The value `5` below is automatically applied to every element.

````python
numbers = np.array([10, 20, 30])

print(numbers + 5)
````

Output:

```text
[15 25 35]
```

### Broadcasting between arrays

````python
col = np.array([[1], [2], [3]])   # shape (3, 1)
row = np.array([10, 20, 30])      # shape (3,)

print(col + row)                  # result shape (3, 3)
````

Broadcasting is an important NumPy concept for Machine Learning and numerical computing.

# 🔗 Joining and ✂️ Splitting Arrays

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate([a, b]))
print(np.stack([a, b]))
print(np.vstack([a, b]))
print(np.hstack([a, b]))
````

````python
arr = np.arange(1, 9)
print(np.split(arr, 4))

m = np.arange(1, 17).reshape(4, 4)
print(np.vsplit(m, 2))
print(np.hsplit(m, 2))
````

---
# 📎 Copy and View

````python
original = np.array([1, 2, 3, 4])

v = original.view()   # shares memory
c = original.copy()   # independent

original[0] = 99

print("original:", original)
print("view    :", v)   # changed too
print("copy    :", c)   # unchanged
````

# 🧮 Linear Algebra

NumPy also provides tools for mathematical and matrix operations, including matrix multiplication, dot product, transpose, determinant, inverse, eigenvalues and eigenvectors.

````python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.matmul(a, b)

print(result)
````

````python
print("Dot product  :", np.dot(np.array([1, 2, 3]), np.array([4, 5, 6])))
print("Determinant  :", np.linalg.det(a))
print("Inverse      :\n", np.linalg.inv(a))
vals, vecs = np.linalg.eig(a)
print("Eigenvalues  :", vals)
print("Eigenvectors :\n", vecs)
````

# 🧹 Handling Numerical Data

NumPy can also be used for basic data preparation. It provides functions that help identify and work with numerical data issues.

````python
data = np.array([10, 20, np.nan, 40, 50])

print(np.isnan(data))
print("Mean ignoring NaN:", np.nanmean(data))

# Replace missing value with the mean of the others
clean = np.where(np.isnan(data), np.nanmean(data), data)
print(clean)
````

### Normalization & Standardization

````python
x = np.array([10., 20., 30., 40., 50.])

normalized   = (x - x.min()) / (x.max() - x.min())   # scale to 0..1
standardized = (x - x.mean()) / x.std()              # mean 0, std 1

print(normalized)
print(standardized)
````
