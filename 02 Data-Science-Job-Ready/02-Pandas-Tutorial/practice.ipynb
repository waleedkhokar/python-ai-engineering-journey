```python
import pandas as pd

df = pd.DataFrame({
    "salary": [10, 12, 13, 14, 15, 16, 17, 18, 50]
})

Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["salary"] < lower) |
    (df["salary"] > upper)
]

print(outliers)
```

```python
from scipy.stats import zscore

df["z_score"] = zscore(df["salary"])

outliers = df[
    df["z_score"].abs() > 3
]

print(outliers)
```

```python
import numpy as np

data = np.array([12, 15, 18, 20, 22, 25, 27, 30, 35, 100])

Q1 = np.quantile(data, 0.25)
Q2 = np.quantile(data, 0.50)
Q3 = np.quantile(data, 0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower boundary:", lower)
print("Upper boundary:", upper)
```
