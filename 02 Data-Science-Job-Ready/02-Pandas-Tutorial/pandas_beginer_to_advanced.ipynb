# 3️⃣ Pandas Data Structures

Pandas mainly provides two important data structures: **Series** and **DataFrame**.

## 📌 Series

A Pandas Series is a one-dimensional labeled data structure. It contains values, an index and a data type.

````python
import pandas as pd

marks = pd.Series([85, 90, 78, 92])

print(marks)
print("Values:", marks.values)
print("Index :", marks.index)
print("dtype :", marks.dtype)
````

## 📊 DataFrame

A DataFrame is a two-dimensional labeled data structure that looks like a table or spreadsheet.

```text
Rows    → Records
Columns → Features / Variables
```

````python
data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [22, 25, 21],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print(df)
````

## 🔄 Series vs DataFrame

| Feature    | Series          | DataFrame        |
| ---------- | --------------- | ---------------- |
| Dimension  | 1D              | 2D               |
| Structure  | Single column   | Multiple columns |
| Similar to | List / column   | Table            |
| Has index  | Yes             | Yes              |
| Common use | Single variable | Complete dataset |

````python
print(type(df["Age"]))   # a single column is a Series
print(type(df))          # the whole table is a DataFrame
````

# 🔎 Selecting Rows

Pandas provides `loc` and `iloc`.

## 📍 `loc` — label-based

````python
print(df.loc[0])
df.loc[0:2, ["Name", "Marks"]]   # NOTE: loc slices INCLUDE the end label
````

## 🔢 `iloc` — position-based

````python
print(df.iloc[0])
display(df.iloc[0:3])        # end position is EXCLUDED
df.iloc[0:3, 0:2]
````

## 🆚 `loc` vs `iloc`

| Method | Selection      |
| ------ | -------------- |
| `loc`  | Label-based    |
| `iloc` | Position-based |

`df.loc[2]` → row with **label** `2`. `df.iloc[2]` → the **third row** by position.

They only differ when the index is not `0,1,2,...` — try it:

````python
tmp = df.set_index("Name")
print(tmp.loc["Sara", "Marks"])   # by label
print(tmp.iloc[2, -1])            # by position
````

## 🔗 Multiple Conditions

Use `&` (and), `|` (or), `~` (not). Always wrap each condition in parentheses.

````python
print("AND:")
display(df[(df["Age"] > 20) & (df["Marks"] > 80)])
print("OR:")
display(df[(df["Age"] > 24) | (df["Marks"] > 90)])
print("NOT:")
df[~(df["Marks"] > 80)]
````

Real-world datasets often contain missing values.


Important methods: `isna()`, `isnull()`, `notna()`, `dropna()`, `fillna()`.

````python
import numpy as np

emp = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [22, np.nan, 24],
    "Salary": [50000, 60000, np.nan]
})
emp
````

# 📅 Working with Dates

Date handling is very important for real-world datasets.

````python
dt = pd.DataFrame({"Date": ["2024-01-15", "2024-02-03", "2024-03-20"]})

dt["Date"] = pd.to_datetime(dt["Date"])

dt["Year"] = dt["Date"].dt.year
dt["Month"] = dt["Date"].dt.month
dt["Day"] = dt["Date"].dt.day
dt["Weekday"] = dt["Date"].dt.day_name()
dt
````

# 📊 GroupBy

`groupby()` is one of the most important Pandas concepts. It allows us to group data based on a category and perform calculations.

````python
employees = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [90000, 110000, 60000, 65000, 70000, 85000],
    "Age": [28, 35, 30, 41, 26, 38],
})

employees.groupby("Department")["Salary"].mean()   # average salary per department
````

# 🔗 Combining DataFrames

Important operations: `concat()`, `merge()`, `join()`.

## ➕ `concat()`

````python
df1 = pd.DataFrame({"Name": ["Ali", "Sara"], "Marks": [85, 88]})
df2 = pd.DataFrame({"Name": ["Hina", "Bilal"], "Marks": [92, 45]})

pd.concat([df1, df2], ignore_index=True)          # vertical
````

````python
pd.concat([df1, df2], axis=1)                     # horizontal
````

## 🔗 `merge()`

Combines datasets using common columns. Similar to a SQL JOIN.

````python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Ali", "Sara", "Hina", "Bilal"],
})
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 5],
    "amount": [250, 100, 400, 75],
})

pd.merge(customers, orders, on="customer_id")     # inner (default)
````

````python
for how in ["inner", "left", "right", "outer"]:
    print(f"--- {how.upper()} ---")
    display(pd.merge(customers, orders, on="customer_id", how=how))
````

## 🆚 Pandas Merge and SQL JOIN

| Pandas        | SQL               |
| ------------- | ----------------- |
| `merge()`     | `JOIN`            |
| `on=`         | `ON`              |
| `how="inner"` | `INNER JOIN`      |
| `how="left"`  | `LEFT JOIN`       |
| `how="right"` | `RIGHT JOIN`      |
| `how="outer"` | `FULL OUTER JOIN` |

## `join()` — merges on the **index**

````python
customers.set_index("customer_id").join(orders.set_index("customer_id"), how="left")
````

# 🧹 Data Cleaning

Data cleaning is one of the most important parts of Data Science.

```text
Raw Dataset → Check Shape → Check Data Types → Check Missing Values → Check Duplicates
→ Fix Incorrect Values → Clean Text → Convert Data Types → Handle Outliers → Clean Dataset
```

A complete example on a deliberately messy dataset:

````python
messy = pd.DataFrame({
    "Name": ["  ali ", "SARA", "Ahmed", "SARA", "hina"],
    "Age": ["22", "21", None, "21", "23"],
    "City": ["islamabad", "LAHORE ", "Karachi", "LAHORE ", None],
    "Marks": [85, 88, 90, 88, 92],
})
print("BEFORE"); display(messy); print(messy.dtypes)

clean = messy.copy()
clean["Name"] = clean["Name"].str.strip().str.title()            # fix text
clean["City"] = clean["City"].str.strip().str.title()
clean["Age"] = pd.to_numeric(clean["Age"])                        # fix dtype
clean["Age"] = clean["Age"].fillna(clean["Age"].median()).astype(int)  # missing
clean["City"] = clean["City"].fillna("Unknown")
clean = clean.drop_duplicates().reset_index(drop=True)            # duplicates

print("AFTER"); display(clean); print(clean.dtypes)
````

# 🚀 Mini Project — Student Performance Analysis

Dataset columns: `Student_ID, Name, Age, Gender, City, Study_Hours, Attendance, Math, Science, English`.

Below is a generated dataset so you can complete each step right away.

### Step 1 — Load Data

````python
np.random.seed(7)
n = 30
project = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 26, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "City": np.random.choice(["Islamabad", "Lahore", "Karachi"], n),
    "Study_Hours": np.random.randint(1, 9, n).astype(float),
    "Attendance": np.random.randint(60, 100, n).astype(float),
    "Math": np.random.randint(30, 100, n).astype(float),
    "Science": np.random.randint(30, 100, n).astype(float),
    "English": np.random.randint(30, 100, n).astype(float),
})
# add some dirt
project.loc[3, "Math"] = np.nan
project.loc[8, "Attendance"] = np.nan
project = pd.concat([project, project.iloc[[0]]], ignore_index=True)   # a duplicate row
project.to_csv("student_performance.csv", index=False)

sp = pd.read_csv("student_performance.csv")
sp.head()
````

### Step 2 — Inspect

Check shape, columns, data types, missing values, duplicates.

````python
print("Shape     :", sp.shape)
print("Columns   :", list(sp.columns))
print(sp.dtypes)
print("Missing   :\n", sp.isna().sum())
print("Duplicates:", sp.duplicated().sum())
````

### Step 3 — Clean

````python
sp = sp.drop_duplicates().reset_index(drop=True)
sp["Math"] = sp["Math"].fillna(sp["Math"].mean())
sp["Attendance"] = sp["Attendance"].fillna(sp["Attendance"].median())
print(sp.isna().sum().sum(), "missing values left")
````

### Step 4 — Transform

Create `Total_Marks`, `Average_Marks`, `Pass_Fail`, `Performance`.

````python
subjects = ["Math", "Science", "English"]
sp["Total_Marks"] = sp[subjects].sum(axis=1)
sp["Average_Marks"] = sp[subjects].mean(axis=1).round(2)
sp["Pass_Fail"] = np.where(sp["Average_Marks"] >= 50, "Pass", "Fail")
sp["Performance"] = pd.cut(
    sp["Average_Marks"],
    bins=[0, 50, 65, 80, 100],
    labels=["Poor", "Average", "Good", "Excellent"]
)
sp.head()
````

### Step 5 — Analyze

````python
print("Average marks       :", round(sp["Average_Marks"].mean(), 2))
print("Highest marks       :", sp["Total_Marks"].max())
print("Lowest marks        :", sp["Total_Marks"].min())
print("Average attendance  :", round(sp["Attendance"].mean(), 2))
print("Average study hours :", round(sp["Study_Hours"].mean(), 2))

print("\nTop 5 students:")
display(sp.nlargest(5, "Total_Marks")[["Name", "Total_Marks", "Average_Marks"]])

print("Students below average:", (sp["Average_Marks"] < sp["Average_Marks"].mean()).sum())

print("\nPerformance by city:")
display(sp.groupby("City")["Average_Marks"].mean().round(2))
print("Performance by gender:")
display(sp.groupby("Gender")["Average_Marks"].mean().round(2))
````

### Step 6 — Prepare for Visualization

Summary tables that can later be plotted with Matplotlib / Seaborn.

````python
city_summary = sp.groupby("City").agg(
    Students=("Student_ID", "count"),
    Avg_Marks=("Average_Marks", "mean"),
    Avg_Attendance=("Attendance", "mean"),
).round(2)

performance_counts = sp["Performance"].value_counts().sort_index()

display(city_summary)
performance_counts
````
