# 📊 First Matplotlib Plot

The simplest line plot:

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)

plt.show()
````

## 🧩 Understanding the Plot

`x` = X-axis values, `y` = Y-axis values. This creates the points `(1,10) (2,20) (3,30) (4,40) (5,50)` and Matplotlib connects them to create the line.

## 🏷️ Adding a Legend

A legend explains what different lines or data series represent.

````python
plt.plot(x, y, label="Sales")

plt.legend()

plt.show()
````

# 📈 Line Charts

Useful for showing trends, changes over time, continuous data, growth and decline.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

## 📈 Multiple Lines

Multiple datasets can be displayed on one chart.

````python
months = ["Jan", "Feb", "Mar", "Apr"]

sales_2023 = [100, 120, 140, 160]
sales_2024 = [120, 150, 170, 200]

plt.plot(months, sales_2023, label="2023")
plt.plot(months, sales_2024, label="2024")

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
````

## 🎨 Line Styles

| Style | Meaning  |
| ----- | -------- |
| `-`   | Solid    |
| `--`  | Dashed   |
| `:`   | Dotted   |
| `-.`  | Dash-dot |

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

for i, style in enumerate(["-", "--", ":", "-."]):
    plt.plot(x, [v + i * 10 for v in y], linestyle=style, label=f"linestyle='{style}'")

plt.legend()
plt.title("Line Styles")
plt.show()
````

## 🔘 Markers

Markers show individual data points. Common markers: `o s ^ v * D x +`

````python
for i, m in enumerate(["o", "s", "^", "v", "*", "D", "x", "+"]):
    plt.plot(x, [v + i * 8 for v in y], marker=m, label=f"marker='{m}'")

plt.legend(ncol=2, fontsize=8)
plt.title("Markers")
plt.show()
````

## 📏 Line Width, 🎨 Colors and Marker Size

You can also use hexadecimal color codes (e.g. `#ff6600`).

````python
plt.plot(x, y, linewidth=1, color="blue", label="linewidth=1")
plt.plot(x, [v + 15 for v in y], linewidth=3, color="#ff6600", marker="o", markersize=10, label="linewidth=3, hex color")
plt.legend()
plt.show()
````

# 📊 Bar Charts

Bar charts are useful for comparing categories.

````python
subjects = ["Math", "Science", "English"]
marks = [85, 90, 78]

plt.bar(subjects, marks)

plt.title("Student Marks")
plt.xlabel("Subject")
plt.ylabel("Marks")

plt.show()
````

## ↔️ Horizontal Bar Chart — `plt.barh()`

````python
plt.barh(subjects, marks)

plt.title("Student Marks")

plt.show()
````

## 📊 Multiple (Grouped) Bar Charts

Grouped bars compare multiple groups side-by-side. Shift each group's x-position by the bar width.

````python
subjects = ["Math", "Science", "English"]

boys = [80, 85, 75]
girls = [90, 88, 82]

pos = np.arange(len(subjects))
width = 0.35

plt.bar(pos - width/2, boys, width, label="Boys")
plt.bar(pos + width/2, girls, width, label="Girls")

plt.xticks(pos, subjects)
plt.title("Average Marks by Gender")
plt.legend()
plt.show()
````

## 📚 Stacked Bar Charts

Stacked bars display both **Total + Composition**.

````python
categories = ["Q1", "Q2", "Q3", "Q4"]
first_values = [20, 35, 30, 35]
second_values = [25, 32, 34, 20]

plt.bar(categories, first_values, label="Product A")

plt.bar(categories, second_values, bottom=first_values, label="Product B")

plt.legend()
plt.title("Stacked Bar Chart")
plt.show()
````

## 🏷️ Bar Labels (value labels on bars)

````python
bars = plt.bar(["Math", "Science", "English"], [85, 90, 78], color=["#4c72b0", "#55a868", "#c44e52"])
plt.bar_label(bars)            # adds the value above each bar
plt.ylim(0, 100)
plt.title("Marks with Value Labels")
plt.show()
````

# 🔵 Scatter Plots

Scatter plots show relationships between two numerical variables.

````python
hours = [1, 2, 3, 4, 5, 6]
marks = [50, 55, 65, 70, 80, 90]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
````

## 🧠 Understanding Scatter Plots

```text
X = Study Hours     Y = Marks
Student → Study Hours + Marks → One point
```

This lets us visually inspect whether higher study hours appear associated with higher marks.

> ⚠️ A visualization can show a pattern, but correlation or causation should not be concluded from a plot alone.

### Size, color mapping and multiple groups

````python
np.random.seed(1)
hours = np.random.uniform(1, 8, 40)
marks = 40 + hours * 6 + np.random.normal(0, 5, 40)
attendance = np.random.uniform(60, 100, 40)

sc = plt.scatter(hours, marks, c=attendance, s=attendance, cmap="viridis", alpha=0.8)
plt.colorbar(sc, label="Attendance %")
plt.title("Study Hours vs Marks (color = attendance)")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
````

````python
# Multiple groups
g1 = np.random.normal(60, 8, 30); h1 = np.random.uniform(1, 5, 30)
g2 = np.random.normal(80, 8, 30); h2 = np.random.uniform(4, 8, 30)

plt.scatter(h1, g1, label="Group A", marker="o")
plt.scatter(h2, g2, label="Group B", marker="^")
plt.legend()
plt.title("Two Groups")
plt.show()
````

# 📊 Histograms

A histogram shows the distribution of numerical data.

````python
ages = [
    18, 19, 20, 20, 21,
    22, 22, 23, 24, 25
]

plt.hist(ages)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
````

## 🪣 Histogram Bins

Bins divide numerical values into ranges. Too few bins → less detail; too many → more detail but potentially noisy.

````python
fig, axes = plt.subplots(1, 3, figsize=(12, 3))
data = np.random.normal(50, 10, 500)
for ax, b in zip(axes, [3, 10, 50]):
    ax.hist(data, bins=b)
    ax.set_title(f"bins={b}")
plt.tight_layout()
plt.show()
````

## 📊 Density Histogram & comparing distributions

Density is useful when comparing distributions with different sample sizes.

````python
a = np.random.normal(50, 10, 1000)
b = np.random.normal(65, 8, 200)

plt.hist(a, bins=20, density=True, alpha=0.6, label="Group A (n=1000)")
plt.hist(b, bins=20, density=True, alpha=0.6, label="Group B (n=200)")
plt.legend()
plt.title("Density Histograms")
plt.show()
````

# 🥧 Pie Charts

Pie charts show parts of a whole.

````python
labels = ["Python", "JavaScript", "Java"]
values = [50, 30, 20]

plt.pie(values, labels=labels)

plt.title("Programming Language Usage")

plt.show()
````

````python
explode = [0.1, 0, 0]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90,
    shadow=True
)
plt.legend(title="Language", loc="center left", bbox_to_anchor=(1, 0.5))
plt.title("Programming Language Usage")
plt.show()
````

# 📦 Box Plots

Box plots summarize the distribution of numerical data and help identify median, quartiles, spread and potential outliers.

````python
marks = [45, 50, 55, 60, 65, 70, 75, 80, 95]

plt.boxplot(marks)

plt.title("Marks Distribution")

plt.show()
````

## 🧠 Box Plot Concepts

```text
Minimum → Q1 → Median → Q3 → Maximum      (IQR = Q3 − Q1)
```

Whiskers extend to the most extreme points within 1.5 × IQR; points beyond are drawn as potential outliers. Below we compare distributions and include an outlier:

````python
class_a = [45, 50, 55, 60, 65, 70, 75, 80, 95]
class_b = [60, 62, 65, 67, 70, 72, 75, 78, 20]   # 20 is an outlier

plt.boxplot([class_a, class_b], tick_labels=["Class A", "Class B"])
plt.title("Comparing Distributions")
plt.show()
````

# 🧩 Subplots

Multiple plots can be displayed in one figure.

```text
┌────────────┬────────────┐
│   Plot 1   │   Plot 2   │
├────────────┼────────────┤
│   Plot 3   │   Plot 4   │
└────────────┴────────────┘
```

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y)
axes[0, 1].bar(x, y)
axes[1, 0].scatter(x, y)
axes[1, 1].hist(y)

plt.tight_layout()

plt.show()
````

## `plt.subplot()` (older style) and sharing axes

````python
plt.subplot(1, 2, 1)     # rows, cols, index
plt.plot(x, y)
plt.title("Left")

plt.subplot(1, 2, 2)
plt.bar(x, y)
plt.title("Right")

plt.tight_layout()
plt.show()
````

````python
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, figsize=(8, 3))
ax1.plot(x, y);  ax1.set_title("Shared Y")
ax2.plot(x, [v * 2 for v in y])
plt.show()
````

# 🧠 Figure and Axes — Object-Oriented Style

A professional Matplotlib workflow often uses `fig, ax = plt.subplots()` and then `ax.plot(x, y)` instead of `plt.plot(x, y)`.

## 🆚 Pyplot vs Object-Oriented Style

### Pyplot Style

````python
plt.plot(x, y)
plt.title("Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

### Object-Oriented Style

````python
fig, ax = plt.subplots()

ax.plot(x, y, label="Sales")
ax.set_title("Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()

plt.show()
````

The object-oriented style becomes especially useful for multiple plots, subplots, complex visualizations, reusable visualization functions, and professional dashboards and reports.

````python
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(["A", "B", "C"], [3, 7, 5])
ax.scatter(["A", "B", "C"], [4, 6, 8], color="red", zorder=3)
ax.set_title("ax.bar() + ax.scatter()")
ax.set_xlabel("Category"); ax.set_ylabel("Value")
plt.show()
````

# 💾 Saving Charts

````python
fig, ax = plt.subplots()
ax.plot([1, 2, 3, 4], [100, 150, 130, 180])
ax.set_title("Sales")

fig.savefig("sales_chart.png")                       # PNG
fig.savefig("sales_chart_hd.png", dpi=300)           # high resolution
fig.savefig("sales_chart.pdf")                       # PDF
fig.savefig("sales_chart.jpg", dpi=150)              # JPG
fig.savefig("sales_transparent.png", transparent=True, bbox_inches="tight")
plt.show()
print("Saved: sales_chart.png / _hd.png / .pdf / .jpg / sales_transparent.png")
````

Higher DPI is useful for high-quality reports. `bbox_inches='tight'` trims extra white space. Matplotlib supports several output formats depending on the backend and file extension.

# 🎯 Choosing the Right Chart

| Goal                       | Common Chart   |
| -------------------------- | -------------- |
| Show trend over time       | Line           |
| Compare categories         | Bar            |
| Compare many categories    | Horizontal bar |
| Show distribution          | Histogram      |
| Show relationship          | Scatter        |
| Show median and spread     | Box plot       |
| Show simple part-to-whole  | Pie            |
| Show cumulative/trend area | Area           |
| Compare multiple views     | Subplots       |

The chart should communicate the data clearly rather than simply making the visualization more complicated.

## 🧠 Visualization Decision Guide

```text
What am I trying to understand? → What type of data? → What is my goal? → Choose a chart

Trend → Line Chart              Category Comparison → Bar Chart
Distribution → Histogram / Box Plot      Relationship → Scatter Plot
Simple Part-to-Whole → Pie Chart
```

Choosing the correct visualization is more important than adding unnecessary styling.

A quick EDA on a generated dataset — histogram, box plot, bar chart and scatter in one figure:

````python
np.random.seed(3)
n = 100
eda = pd.DataFrame({
    "Study_Hours": np.random.uniform(1, 8, n),
    "Gender": np.random.choice(["Male", "Female"], n),
})
eda["Marks"] = (40 + eda["Study_Hours"] * 6 + np.random.normal(0, 6, n)).clip(0, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].hist(eda["Marks"], bins=15)
axes[0, 0].set_title("Distribution of Marks")

axes[0, 1].boxplot(eda["Marks"])
axes[0, 1].set_title("Marks Box Plot (spread & outliers)")

eda.groupby("Gender")["Marks"].mean().plot(kind="bar", ax=axes[1, 0], rot=0)
axes[1, 0].set_title("Average Marks by Gender")

axes[1, 1].scatter(eda["Study_Hours"], eda["Marks"], alpha=0.7)
axes[1, 1].set_title("Study Hours vs Marks")

plt.tight_layout()
plt.show()
````

## 📊 Example — Training Loss

````python
epochs = [1, 2, 3, 4, 5]
loss = [0.8, 0.6, 0.45, 0.35, 0.28]

plt.plot(epochs, loss)

plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.show()
````

This allows us to visually inspect how loss changes during training. A common pattern is plotting training and validation loss together:

````python
val_loss = [0.85, 0.65, 0.55, 0.5, 0.5]
plt.plot(epochs, loss, marker="o", label="Training loss")
plt.plot(epochs, val_loss, marker="s", label="Validation loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.legend(); plt.grid(alpha=0.3)
plt.title("Training vs Validation Loss")
plt.show()
````

```text
NumPy → Numerical Computing → Pandas → Data Manipulation & Analysis → Matplotlib → Data Visualization

Raw Data → Pandas → Clean Data → NumPy → Numerical Operations → Matplotlib → Visualization → Insights
```

````python
# All three together
np.random.seed(5)
raw = pd.DataFrame({"Day": range(1, 31), "Sales": np.random.randint(100, 300, 30)})   # Pandas
raw["Trend"] = np.convolve(raw["Sales"], np.ones(5) / 5, mode="same")                # NumPy

plt.plot(raw["Day"], raw["Sales"], alpha=0.5, label="Daily sales")                   # Matplotlib
plt.plot(raw["Day"], raw["Trend"], linewidth=3, label="5-day moving average")
plt.legend(); plt.xlabel("Day"); plt.ylabel("Sales"); plt.title("NumPy + Pandas + Matplotlib")
plt.show()
````

# 🚀 Mini Project — Student Performance Visualization

Dataset columns: `Student_ID, Name, Age, Gender, Study_Hours, Attendance, Math, Science, English` (+ a `Month` column for the trend chart). A generated dataset is provided so you can start right away.

````python
np.random.seed(11)
n = 60
sp = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 26, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Study_Hours": np.random.uniform(1, 8, n).round(1),
    "Attendance": np.random.uniform(60, 100, n).round(1),
})
base = 35 + sp["Study_Hours"] * 5
for subj in ["Math", "Science", "English"]:
    sp[subj] = (base + np.random.normal(0, 8, n)).clip(0, 100).round(1)
sp["Average"] = sp[["Math", "Science", "English"]].mean(axis=1)
sp["Month"] = np.random.choice(["Jan", "Feb", "Mar", "Apr", "May"], n)
sp.head()
````

### 1. Marks Distribution

````python
plt.hist(sp["Average"], bins=12, edgecolor="black")
plt.title("Distribution of Average Marks"); plt.xlabel("Average Marks"); plt.ylabel("Students")
plt.show()
````

### 2. Subject Comparison

````python
avg = sp[["Math", "Science", "English"]].mean()
bars = plt.bar(avg.index, avg.values, color=["#4c72b0", "#55a868", "#c44e52"])
plt.bar_label(bars, fmt="%.1f")
plt.title("Average Marks by Subject"); plt.ylabel("Average Marks")
plt.show()
````

### 3. Study Hours vs Marks

````python
plt.scatter(sp["Study_Hours"], sp["Average"], alpha=0.7)
plt.title("Study Hours vs Average Marks"); plt.xlabel("Study Hours"); plt.ylabel("Average Marks")
plt.show()
````

### 4. Attendance Distribution

````python
plt.hist(sp["Attendance"], bins=10, edgecolor="black", color="orange")
plt.title("Attendance Distribution"); plt.xlabel("Attendance %"); plt.ylabel("Students")
plt.show()
````

### 5. Gender Comparison

````python
sp.groupby("Gender")["Average"].mean().plot(kind="bar", rot=0, color=["#dd8452", "#4c72b0"])
plt.title("Average Marks by Gender"); plt.ylabel("Average Marks")
plt.show()
````

### 6. Student Performance (Box Plot)

````python
plt.boxplot([sp["Math"], sp["Science"], sp["English"]], tick_labels=["Math", "Science", "English"])
plt.title("Marks Distribution by Subject")
plt.show()
````

### 7. Monthly / Sequential Trend

````python
order = ["Jan", "Feb", "Mar", "Apr", "May"]
monthly = sp.groupby("Month")["Average"].mean().reindex(order)
plt.plot(monthly.index, monthly.values, marker="o")
plt.title("Average Marks by Month"); plt.xlabel("Month"); plt.ylabel("Average Marks"); plt.grid(alpha=0.3)
plt.show()
````

### 8. Dashboard-Style Figure

````python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
fig.suptitle("Student Performance Dashboard", fontsize=16)

axes[0, 0].hist(sp["Average"], bins=12, edgecolor="black"); axes[0, 0].set_title("Marks Distribution")
axes[0, 1].bar(avg.index, avg.values, color=["#4c72b0", "#55a868", "#c44e52"]); axes[0, 1].set_title("Subject Averages")
axes[0, 2].scatter(sp["Study_Hours"], sp["Average"], alpha=0.7); axes[0, 2].set_title("Study Hours vs Marks")
axes[1, 0].hist(sp["Attendance"], bins=10, color="orange", edgecolor="black"); axes[1, 0].set_title("Attendance")
axes[1, 1].boxplot([sp["Math"], sp["Science"], sp["English"]], tick_labels=["Math", "Sci", "Eng"]); axes[1, 1].set_title("Box Plot")
axes[1, 2].plot(monthly.index, monthly.values, marker="o"); axes[1, 2].set_title("Monthly Trend")

plt.tight_layout()
fig.savefig("student_dashboard.png", dpi=200)
plt.show()
````

The goal is to turn a raw dataset into meaningful visual information.
