```python
import matplotlib.pyplot as plt
import numpy as np

departments = ["IT", "HR", "Finance"]

male = [30, 15, 12]
female = [20, 15, 8]

x = np.arange(len(departments))

width = 0.35

plt.bar(x - width/2, male, width, label="Male")
plt.bar(x + width/2, female, width, label="Female")

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(x, departments)

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

classes = ["Cat", "Dog", "Bird"]

training = [1000, 1200, 800]
testing = [200, 250, 150]

x = np.arange(len(classes))

width = 0.35

plt.bar(x - width/2, training, width, label="Training")
plt.bar(x + width/2, testing, width, label="Testing")

plt.title("Training vs Testing Samples")
plt.xlabel("Class")
plt.ylabel("Number of Samples")

plt.xticks(x, classes)

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ["IT", "HR", "Finance"]
employees = [50, 30, 20]

bars = plt.bar(departments, employees)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Employees")

plt.bar_label(bars)

plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ["IT", "HR", "Finance"]
employees = [50, 30, 20]

bars = plt.bar(
    departments,
    employees,
    width=0.5
)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Employees")

plt.bar_label(bars)

plt.grid(axis="y")

plt.show()
```
