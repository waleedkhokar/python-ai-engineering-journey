## ❓ 2. Why Do We Need NLP?

A computer does **not** naturally understand:

```text
"I really love this laptop."
```

the way a human does.

For an ML model, text must eventually become **numerical data**:

```text
"I love Python"
        ↓
Tokenization
        ↓
["I", "love", "Python"]
        ↓
Vectorization
        ↓
[0, 1, 1, 0, ...]
        ↓
Machine Learning Model
```

This conversion from language into **useful numerical representations** is one of the central ideas of NLP.

````python
# Text → tokens → IDs (a tiny illustration of the idea)
text = "I love Python"

# Step 1: tokenize (simple whitespace split)
tokens = text.split()

# Step 2: build a tiny vocabulary
vocab = {tok: i for i, tok in enumerate(sorted(set(tokens)))}

# Step 3: map tokens → IDs (a crude "vector")
ids = [vocab[t] for t in tokens]

print('Text    :', text)
print('Tokens  :', tokens)
print('Vocab   :', vocab)
print('IDs     :', ids)
print()
print('This is the essence of converting text to numbers.')
````

## 🧩 3. Why Is Human Language Difficult?

Human language is **highly variable**:

```text
"I love this phone."
"I don't love this phone."
"I LOVE this phone!!!"
"This phone is sick 🔥"
```

The same words can have **different meanings** depending on:

- Context
- Word order
- Negation
- Grammar
- Punctuation
- Slang
- Spelling
- Domain
- Tone

For example:

```text
"I like this."
"I don't like this."
```

The word **like** is present in both — but **not** changes the meaning completely.

> NLP cannot simply treat every word independently.

````python
# Same words, very different meanings
sentences = [
    "I love this phone.",
    "I don't love this phone.",
    "I LOVE this phone!!!",
    "This phone is sick 🔥",
]

for i, s in enumerate(sentences, 1):
    print(f'{i}. {s}')

print()
print('Words alone are not enough — context changes everything.')
````

## 📝 8. Raw Text

Raw text is the original **unprocessed** language data.

Example:

```text
"Wow!!! This phone is AMAZING 😍.
I bought it yesterday."
```

May contain:

```text
Uppercase letters
Punctuation
Emoji
Whitespace
Multiple sentences
```

Before classical ML, we may transform the text into a consistent representation.

But we should **not blindly remove information**:

```text
"Great."
"Great!!!"
```

may contain **different levels of emotional emphasis**.

````python
# Examine raw text
raw = "Wow!!! This phone is AMAZING 😍. I bought it yesterday."

print('Raw text:', raw)
print()
print('Length (chars):', len(raw))
print('Words         :', raw.split())
print('Has uppercase :', any(c.isupper() for c in raw))
print('Has punctuation:', bool(re.search(r'[!.,?]', raw)))
print('Has emoji     :', any(ord(c) > 0xFFFF for c in raw))
````

## 📖 11. Important NLP Vocabulary

### Corpus
A collection of text documents.

```text
Corpus
 ├── Document 1
 ├── Document 2
 ├── Document 3
 └── Document 4
```

### Document
A single piece of text.

```text
"This is one document."
```

### Token
A smaller unit extracted from text.

```text
"I love Python"
→ ["I", "love", "Python"]
```

### Vocabulary
The collection of **unique** tokens in a corpus.

```text
["I", "love", "Python", "machine", "learning"]
```

> These terms become important when we learn vectorization.

````python
# Corpus → Documents → Tokens → Vocabulary
corpus = [
    "I love Python",
    "I love machine learning",
    "Python is powerful",
]

print('Corpus (documents):')
for i, doc in enumerate(corpus, 1):
    print(f'  Doc {i}: {doc}')

# Tokenize (simple)
tokenized = [doc.lower().split() for doc in corpus]

print('\nTokens per document:')
for i, toks in enumerate(tokenized, 1):
    print(f'  Doc {i}: {toks}')

# Build vocabulary
all_tokens = [tok for doc in tokenized for tok in doc]
vocab = sorted(set(all_tokens))

print('\nVocabulary (unique tokens):', vocab)
print('Vocabulary size            :', len(vocab))
````

````python
# Frequency of each token in the corpus
freq = Counter(all_tokens)

plt.figure(figsize=(8, 3.5))
plt.bar(list(freq.keys()), list(freq.values()))
plt.title('Token frequency in the corpus')
plt.xlabel('Token'); plt.ylabel('Count')
plt.grid(alpha=0.3, axis='y')
plt.tight_layout(); plt.show()

print('Token counts:')
for tok, cnt in freq.most_common():
    print(f'  {tok:<12} {cnt}')
````

## 🎤 18. Interview Questions

1. **What is NLP?**  
   Natural Language Processing — enabling computers to work with human language.

2. **Why is NLP difficult?**  
   Language is ambiguous, contextual, and full of slang, negation, and domain-specific meaning.

3. **NLP vs traditional ML?**  
   NLP must convert unstructured text into numerical features before ML models can be applied.

4. **List main NLP tasks.**  
   Text classification, sentiment, spam, topic modeling, NER, similarity, search, clustering, information extraction.

5. **What is the classical NLP pipeline?**  
   Raw text → Cleaning → Normalization → Tokenization → Feature Extraction → Vectorization → ML → Evaluation → Prediction.

6. **What is a corpus?**  
   A collection of text documents.

7. **What is a document?**  
   A single piece of text.

8. **What is a token?**  
   A smaller unit extracted from text — typically a word, subword, or character.

9. **What is a vocabulary?**  
   The set of unique tokens in the corpus.

10. **What is the representation problem?**  
    ML models need numbers, so text must be converted into numerical vectors.

11. **Why does context matter?**  
    The meaning of a word often depends on surrounding words (e.g., "bank").

12. **What is the bridge to modern NLP?**
