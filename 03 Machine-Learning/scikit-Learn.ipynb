```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Feature (X): Temperature in Celsius (Must be a 2D array: [[val1], [val2]])
X = np.array([[20], [30], [40]])

# Target (y): Ice creams sold
y = np.array([200, 300, 400])

# Step 1: Create the model
model = LinearRegression()

# Step 2: Train the model (Find the pattern line)
model.fit(X, y)

# Step 3: Predict sales for a 35°C day
tomorrow_temp = np.array([[35]])
predicted_sales = model.predict(tomorrow_temp)

print(f"Predicted Ice Cream Sales: {predicted_sales[0]:.0f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([500, 1000, 1500, 2000])
y = np.array([5, 10, 15, 20])

# Linear equation
b0 = 0
b1 = 0.01

# Prediction
y_pred = b0 + b1 * x

print(y_pred)

# Plot
plt.scatter(x, y)
plt.plot(x, y_pred)
plt.xlabel("House Size")
plt.ylabel("Price")
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Training data
X = np.array([[800],
              [1000],
              [1200],
              [1500],
              [1800]])

y = np.array([8, 10, 12, 15, 18])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# New house
new_house = np.array([[800]])

# Prediction
prediction = model.predict(new_house)

print("Predicted price:", prediction)
```
