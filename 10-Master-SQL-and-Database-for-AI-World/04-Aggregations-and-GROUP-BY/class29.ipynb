## Aggregates

```python
%%sql
SELECT COUNT(*), COUNT(phone), COUNT(DISTINCT country) FROM customers;
```

```python
%%sql
SELECT ROUND(AVG(salary),2), MIN(salary), MAX(salary) FROM employees;
```

## GROUP BY / HAVING

```python
%%sql
SELECT department_id, COUNT(*) AS headcount, ROUND(AVG(salary),0) AS avg_salary
FROM employees GROUP BY department_id ORDER BY avg_salary DESC;
```

```python
%%sql
SELECT department_id, AVG(salary) FROM employees GROUP BY department_id HAVING AVG(salary) > 100000;
```

## CASE and conditional aggregation

```python
%%sql
SELECT first_name, salary,
 CASE WHEN salary>=150000 THEN 'High' WHEN salary>=80000 THEN 'Medium' ELSE 'Low' END AS level
FROM employees;
```

```python
%%sql
SELECT COUNT(*) FILTER (WHERE status='paid') AS paid,
       COUNT(*) FILTER (WHERE status='shipped') AS shipped,
       COUNT(*) FILTER (WHERE status='cancelled') AS cancelled
FROM orders;
```

```python
%%sql
SELECT DATE_TRUNC('month', order_date) AS month, COUNT(*) FROM orders GROUP BY 1 ORDER BY 1;
```
