## SELECT and expressions

```python
%%sql
SELECT first_name, salary, salary*12 AS yearly FROM employees;
```

```python
%%sql
SELECT DISTINCT department_id FROM employees;
```

## WHERE

```python
%%sql
SELECT * FROM employees WHERE department_id = 1 AND salary >= 100000;
```

```python
%%sql
SELECT * FROM employees WHERE salary BETWEEN 80000 AND 150000 ORDER BY salary DESC;
```

```python
%%sql
SELECT * FROM products WHERE name ILIKE '%book%';
```

## NULL

```python
%%sql
SELECT name, COALESCE(phone,'N/A') AS phone FROM customers;
```

```python
%%sql
SELECT * FROM customers WHERE phone IS NULL;
```

```python
%%sql
SELECT * FROM customers WHERE phone = NULL;  -- returns nothing!
```

## Functions

```python
%%sql
SELECT UPPER(first_name), LENGTH(email), EXTRACT(YEAR FROM hire_date) AS yr, AGE(hire_date) FROM employees;
```

## Pagination

```python
%%sql
SELECT * FROM employees ORDER BY id LIMIT 5 OFFSET 5;
```

```python
%%sql
-- Keyset pagination
SELECT * FROM employees WHERE id > 5 ORDER BY id LIMIT 5;
```
