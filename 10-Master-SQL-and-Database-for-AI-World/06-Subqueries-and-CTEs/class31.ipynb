## Scalar & correlated subqueries

```python
%%sql
SELECT * FROM employees WHERE salary > (SELECT AVG(salary) FROM employees);
```

```python
%%sql
SELECT e.first_name, e.salary FROM employees e
WHERE e.salary > (SELECT AVG(salary) FROM employees WHERE department_id = e.department_id);
```

## EXISTS / NOT EXISTS

```python
%%sql
SELECT name FROM customers c WHERE EXISTS (SELECT 1 FROM orders o WHERE o.customer_id=c.id AND o.status='paid');
```

```python
%%sql
SELECT name FROM customers c WHERE NOT EXISTS (SELECT 1 FROM orders o WHERE o.customer_id=c.id);
```

## CTE

```python
%%sql
WITH order_totals AS (
  SELECT order_id, SUM(quantity*unit_price) AS total FROM order_items GROUP BY order_id),
paid AS (
  SELECT o.id, o.customer_id, t.total FROM orders o JOIN order_totals t ON t.order_id=o.id WHERE o.status='paid')
SELECT c.name, SUM(p.total) AS revenue FROM paid p JOIN customers c ON c.id=p.customer_id
GROUP BY c.name ORDER BY revenue DESC;
```

## Set operations

```python
%%sql
SELECT id FROM customers INTERSECT SELECT customer_id FROM orders;
```

```python
%%sql
SELECT id FROM customers EXCEPT SELECT customer_id FROM orders;
```

```python
%%sql
SELECT country FROM customers UNION SELECT 'Pakistan' ORDER BY 1;
```
