## 📌 What is Deep Learning?
**Deep Learning** is a subfield of Machine Learning that uses neural networks with multiple computational layers to learn useful representations directly from data.

```text
Traditional Programming:  Rules + Data → Output
Machine Learning:         Data + Features + Algorithm → Model → Output
Deep Learning:            Raw Data → Neural Network → Learned Representations → Prediction
```

## 🧠 AI vs ML vs Deep Learning

```text
Artificial Intelligence (Broad field)
│
└── Machine Learning (Learning patterns from data)
    │
    └── Deep Learning (Multi-layer neural networks)
```

# 1. 🔢 Tensors and Shapes Exploration

In PyTorch, data and parameters are represented using **tensors**. Let's examine scalar, vector, matrix, and image batch tensor representations.

````python
import torch

# 0D Tensor (Scalar)
scalar = torch.tensor(5)
print("Scalar:", scalar, "| Shape:", scalar.shape, "| Dim:", scalar.ndim)

# 1D Tensor (Vector)
vector = torch.tensor([1.0, 2.0, 3.0])
print("Vector:", vector, "| Shape:", vector.shape, "| Dim:", vector.ndim)

# 2D Tensor (Matrix)
matrix = torch.tensor([[1, 2], [3, 4]])
print("Matrix:\n", matrix, "\nShape:", matrix.shape, "| Dim:", matrix.ndim)

# 3D Tensor (Single RGB Image: Channels x Height x Width)
single_image = torch.randn(3, 224, 224)
print("Single Image Shape:", single_image.shape)

# 4D Tensor (Batch of RGB Images: Batch x Channels x Height x Width)
image_batch = torch.randn(32, 3, 224, 224)
print("Image Batch Shape:", image_batch.shape)
````

# 2. 🧮 What a Neuron Learns: Weighted Sum & Activation

A single neuron computes: 

$$z = \sum (w_i x_i) + b$$
$$a = \text{ReLU}(z)$$

````python
# Inputs x = [x1, x2, x3]
x = torch.tensor([1.5, -2.0, 3.0])

# Learnable weights w and bias b
w = torch.tensor([0.4, -0.5, 0.8])
b = torch.tensor(0.1)

# Linear weighted sum z
z = torch.dot(w, x) + b
print("Linear Sum (z):", z.item())

# Activation function (ReLU: max(0, z))
a = torch.relu(z)
print("Activated Output (a):", a.item())
````

# 3. 🧪 Simple PyTorch Neural Network Architecture

Building a simple neural network using PyTorch's `nn.Sequential` module.

````python
import torch.nn as nn

# Define simple feedforward model: 2 Inputs -> 4 Hidden Neurons (ReLU) -> 1 Output
model = nn.Sequential(
    nn.Linear(in_features=2, out_features=4),
    nn.ReLU(),
    nn.Linear(in_features=4, out_features=1)
)

# Pass dummy input through the forward pass
x_in = torch.tensor([[1.0, 2.0]])
output = model(x_in)

print("Model Architecture:\n", model)
print("\nForward Output:", output)
````

# 4. ⚙️ Parameters vs Hyperparameters Inspection

Let's list learnable parameters inside our model vs configurable hyperparameters.

````python
print("--- LEARNABLE PARAMETERS (Updated during training) ---")
total_params = 0
for name, param in model.named_parameters():
    print(f"Parameter: {name:<18} | Shape: {list(param.shape)} | Numel: {param.numel()}")
    total_params += param.numel()

print(f"\nTotal Learnable Parameters: {total_params}")

print("\n--- HYPERPARAMETERS (Configured by Developer) ---")
learning_rate = 0.001
batch_size = 32
epochs = 10
print(f"Learning Rate: {learning_rate} | Batch Size: {batch_size} | Epochs: {epochs}")
````

# 5. 🔁 The Fundamental Deep Learning Training Loop

Below is a complete simulation of a single training step: **Forward Pass → Loss Calculation → Backpropagation → Parameter Update**.

````python
import torch.optim as optim

# 1. Setup Data & Targets
X = torch.tensor([[1.0, 2.0], [2.0, 3.0], [3.0, 1.0]], dtype=torch.float32)
y_target = torch.tensor([[5.0], [8.0], [6.0]], dtype=torch.float32)

# 2. Setup Model, Loss Function, and Optimizer
net = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)
criterion = nn.MSELoss()
optimizer = optim.SGD(net.parameters(), lr=0.01)

print("Initial Target Values:\n", y_target.squeeze())

# 3. Run 5 Simulated Training Steps
for epoch in range(1, 6):
    optimizer.zero_grad()       # Clear previous gradients
    predictions = net(X)        # Forward pass
    loss = criterion(predictions, y_target) # Calculate loss
    loss.backward()             # Backward pass (calculate gradients)
    optimizer.step()            # Update parameters
    
    print(f"Epoch {epoch}/5 - Loss: {loss.item():.4f}")
````
