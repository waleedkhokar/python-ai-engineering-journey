## 4. Verify PyTorch Installation

Now let's make sure PyTorch is installed and working correctly.

````python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("CUDA device count:", torch.cuda.device_count())

# Apple Silicon (MPS) check (macOS only; safe to call on other platforms)
mps_available = hasattr(torch.backends, "mps") and torch.backends.mps.is_available()
print("Apple MPS available:", mps_available)

if torch.cuda.is_available():
    print("Current CUDA device:", torch.cuda.current_device())
    print("Device name:", torch.cuda.get_device_name(0))

# Choose a default device
if torch.cuda.is_available():
    device = torch.device("cuda")
elif mps_available:
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("Using device:", device)
````

## 5. Work with Tensors and Devices

Let's create some tensors on CPU and move them to the selected device. This is the core pattern you'll use in all PyTorch projects:

```python
model.to(device)
tensor.to(device)
```

````python
# Create a tensor on CPU by default
x_cpu = torch.randn(3, 4)
print("x_cpu device:", x_cpu.device)

# Move tensor to our chosen device
x = x_cpu.to(device)
print("x device after .to(device):", x.device)

# Simple operation on the chosen device
w = torch.randn(4, 2, device=device)
y = x @ w
print("y shape:", y.shape)
print("y device:", y.device)
````

### Common Device Mismatch Error

A very common error in PyTorch is mixing CPU and GPU tensors in the same operation.

For example (only runs if CUDA is available):

````python
if torch.cuda.is_available():
    cpu_tensor = torch.randn(2, 2, device="cpu")
    gpu_tensor = torch.randn(2, 2, device="cuda")
    
    try:
        _ = cpu_tensor + gpu_tensor
    except RuntimeError as e:
        print("Expected device mismatch error:")
        print(e)
else:
    print("CUDA not available; skipping mismatch demo.")
````

To avoid this, always:

1. Decide on a `device` at the top of your script/notebook:

   ```python
   device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
   ```

2. Move **both** model and data to that device:

   ```python
   model.to(device)
   inputs = inputs.to(device)
   labels = labels.to(device)
   ```

## 7. Random Seeds and Reproducibility

For educational purposes, it's helpful to make results **reproducible** by fixing random seeds.

Note: Full determinism can be hard across all hardware/OS combinations, but setting seeds is still very useful.

````python
import random
import numpy as np

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Seeds set to:", SEED)

# Small demo: random numbers will be the same each time you run this cell (in the same process)
print("Random tensor:", torch.randn(3, 3))
````
