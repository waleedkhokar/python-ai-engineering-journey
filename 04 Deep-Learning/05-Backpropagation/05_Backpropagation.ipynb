## 1. 📌 Mathematical Foundations
Backpropagation applies the chain rule backward through the computational graph: $\frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w}$

## 2. 🧮 Pure NumPy Forward & Manual Backward Pass

```python
import numpy as np

x, w, b, y_target = 2.0, 3.0, 1.0, 5.0

# Forward Pass
z = w * x + b
loss = (z - y_target) ** 2
print(f"Forward Pass -> z: {z:.2f}, Loss: {loss:.2f}")

# Manual Backward Pass (Chain Rule)
dL_dz = 2.0 * (z - y_target)
dL_dw = dL_dz * x
dL_db = dL_dz * 1.0
print(f"Backward Pass -> dL/dw: {dL_dw:.2f}, dL/db: {dL_db:.2f}")

# Parameter Update
lr = 0.1
w = w - lr * dL_dw
b = b - lr * dL_db
print(f"Updated Params -> w: {w:.2f}, b: {b:.2f}")
```

## 3. 🧪 PyTorch Autograd Equivalence

```python
import torch

x_t = torch.tensor(2.0)
w_t = torch.tensor(3.0, requires_grad=True)
b_t = torch.tensor(1.0, requires_grad=True)
y_t = torch.tensor(5.0)

z_t = w_t * x_t + b_t
loss_t = (z_t - y_t) ** 2
loss_t.backward()

print("PyTorch w.grad:", w_t.grad.item())
print("PyTorch b.grad:", b_t.grad.item())
```

## 4. ⚙ Gradient Accumulation & Inspection

```python
import torch.nn as nn
import torch.optim as optim

model = nn.Linear(2, 1)
optimizer = optim.SGD(model.parameters(), lr=0.01)
criterion = nn.MSELoss()

optimizer.zero_grad()
for i in range(4):
    x_dummy = torch.randn(4, 2)
    y_dummy = torch.randn(4, 1)
    loss = criterion(model(x_dummy), y_dummy) / 4
    loss.backward()

nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
optimizer.step()
optimizer.zero_grad()
print("Accumulation and gradient step complete.")
```
