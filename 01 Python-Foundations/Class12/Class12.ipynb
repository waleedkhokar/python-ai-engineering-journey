# self error generator 

```python
class StudentCard():
    def __init__(self, roll_no: int, name:str, age:int)->None:
        if age > 60 and age < 18:
            raise StudentAgeError("you are not eligible for this ")
        self.roll_no = roll_no
        self.name = name
        self.age = age
        
class StudentAgeError(Exception):
    pass
student1 = StudentCard(12, "Waleed", 20)
print(student1.name, student1.roll_no, student1.age )        
```
